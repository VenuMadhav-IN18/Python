```python
# recursive factorial
def fact(n):
  if n==1:
    return 1
  return n*fact(n-1)
fact(5)
```

```python
def calculator(a,b):
  def addition():
    return a+b
  def subtraction():
    return a+b
  def multiplication():
    return a+b
  def division():
    return a+b
  return addition,subtraction,multiplication,division
```

```python
class calculator:

  def addition(self,a,b):
    return a+b
  def subtraction(self,a,b):
    return a-b
  def multiplication(self,a,b):
    return a*b
  def devision(self,a,b):
    return a/b

cal=calculator()
print(cal.addition(1,2))
print(cal.subtraction(1,2))
print(cal.multiplication(1,2))
print(cal.devision(1,2))
```
